```python
from datetime import datetime
import numpy as np

# Reference time (t0)
dates_t0 = [
    datetime(2023, 1, 1),  # Co
    datetime(2023, 1, 1),  # Na
    datetime(1997, 3, 1),  # Cs
    datetime(1997, 3, 1),  # Ba
    datetime(1997, 3, 1)   # Am
]

# Experiment time
data_actual = datetime(2026, 9, 10)


# Years that have passed
anys_transcorreguts = []
for d in dates_t0:
    diferencia_dies = (data_actual - d).days
    anys_precisos = diferencia_dies / 365.24219878  # Considera anys de traspàs
    anys_transcorreguts.append(anys_precisos)

t_transcorregut = np.array(anys_transcorreguts)


# Half-lives
t_12 = np.array([5.2711, 2.6029, 30.018, 10.539, 432.6])  # Co Na Cs Ba Am
a0 = np.array([157, 28.7, 41.9, 45.9, 38.0])            # kBq

# Current activity
a = a0 * np.exp(-(np.log(2) / t_12) * t_transcorregut)


elements = ["Cobalt-60", "Sodi-22", "Cesi-137", "Bari-133","Americi-241"]
print("Years elapsed:", np.round(t_transcorregut, 3))
print("-" * 50)
for i in range(len(elements)):
    print(f"Current activity of {elements[i]}: {a[i]:.4f} kBq")


# Uncertainties
lam = np.log(2) / t_12

delta_time = np.array([1, 1, 1, 1, 1]) / 365.24219878
delta_t_12 = np.array([0.0008, 0.0008, 0.022, 0.006, 0.6])
delta_a0   = np.array([2, .3, 1.1, 1.5, 1.])

delta_a = np.sqrt( (np.exp(-lam * t_transcorregut) * delta_a0)**2                   # A0
                 + (a * lam * t_transcorregut / t_12 * delta_t_12)**2  )             # half-life


print(a)
print(delta_a)

```

```python
area = [array[0] for array in area_ph]
delta_area = [array[1] for array in area_ph]

activity= [1000*a[0],1000*a[0],1000*a[2],1000*a[1],1000*a[3],1000*a[3],1000*a[4]]
delta_activity = [1000*delta_a[0],1000*delta_a[0],1000*delta_a[2],1000*delta_a[1],1000*delta_a[3],1000*delta_a[3],1000*delta_a[4]]

time = [900,900,1200,3900,1500,1500,1500]
delta_time = [3,3,3,3,3,3,3]

branching_ratio = [.9985,.999826,.8501,.9994,.6205,0.3331,.3592]
delta_branching_ratio = [.0003,.000006,.002,.0013,.0019,.0030,.0017]

distance = [3.73,3.73,3.73,3.73,3.73,3.73,3.73]
delta_distance = [.2,.2,.2,.2,.2,.2,.2]

abs_eff, delta_abs_eff, int_eff, delta_int_eff = analysis.energy_eff(area,delta_area,activity,delta_activity,time,delta_time,branching_ratio,delta_branching_ratio,distance,delta_distance)


energy = np.array([1173.228,1332.492,661.655,1274.537,356.0129,80.9979,59.5409])

print(100*abs_eff)
print(100*delta_abs_eff)
print(energy)
print(100*int_eff)
print(100*delta_int_eff)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
from matplotlib.lines import Line2D
from matplotlib.ticker import AutoMinorLocator, MaxNLocator

# Data (in %), sorted by energy
energy_arr = np.array(energy)
abs_eff_pct = 100 * np.array(abs_eff)
delta_abs_eff_pct = 100 * np.array(delta_abs_eff)

sim_abs_eff_pct = np.array([0.026828, 0.022399, 0.064613, 0.023853, 0.186332, 0.689124, 0.679959])

sort_indices = np.argsort(energy_arr)
energy_sorted = energy_arr[sort_indices]
abs_eff_sorted = abs_eff_pct[sort_indices]
delta_abs_eff_sorted = delta_abs_eff_pct[sort_indices]
sim_abs_eff_pct_sorted = sim_abs_eff_pct[sort_indices]

# Weighted fit: ln(eff) = a*ln(E)^2 + b*ln(E) + c
log_E = np.log(energy_sorted)
log_eff = np.log(abs_eff_sorted)
sigma_log_eff = delta_abs_eff_sorted / abs_eff_sorted

(a, b, c), cov = np.polyfit(log_E, log_eff, deg=2, w=1/sigma_log_eff, cov='unscaled')
da, db, dc = np.sqrt(np.diag(cov))

print("Fit: ln(eff) = a*ln(E)^2 + b*ln(E) + c")
print(f"  a = {a:+.5f} +/- {da:.5f}")
print(f"  b = {b:+.5f} +/- {db:.5f}")
print(f"  c = {c:+.5f} +/- {dc:.5f}")

# Fit curve
energy_dense = np.linspace(energy_sorted.min(), energy_sorted.max(), 200)
log_E_dense = np.log(energy_dense)
eff_fit = np.exp(a * log_E_dense**2 + b * log_E_dense + c)

# Residuals (simulation vs experiment, in units of sigma_exp)
residuals = (abs_eff_sorted - sim_abs_eff_pct_sorted) / delta_abs_eff_sorted

# Markers (same marker for lines of the same source)
shared_markers = {
    'o': [80, 356],
    's': [1173, 1332],
}
other_markers = iter(['^', 'D', 'v', 'p', 'h', '*', 'X', '<', '>'])

markers = []
for E in energy_sorted:
    for m, E_list in shared_markers.items():
        if any(abs(E - E0) < 5 for E0 in E_list):
            markers.append(m)
            break
    else:
        markers.append(next(other_markers))

open_style = dict(markerfacecolor='none', markeredgewidth=1.2, markersize=9,
                  capsize=4, linestyle='none', elinewidth=1)



# Figure: efficiency (top) and residuals (bottom)
fig, (ax1, ax2) = plt.subplots(
    2, 1, sharex=True, figsize=(10, 8),
    gridspec_kw={'height_ratios': [3, 1]},
)
fig.set_layout_engine('constrained', hspace=0.02)

# Top panel
for E, eff, deff, m in zip(energy_sorted, abs_eff_sorted, delta_abs_eff_sorted, markers):
    ax1.errorbar(E, eff, yerr=deff, fmt=m, color='black', **open_style)

fit_line, = ax1.plot(energy_dense, eff_fit, '-', color='crimson', linewidth=2)

sim_pts, = ax1.plot(energy_sorted, sim_abs_eff_pct_sorted, 'x',
                    color='#0072B2', markersize=7, markeredgewidth=2,
                    linestyle='none', zorder=2)

unique_markers = list(dict.fromkeys(markers))
exp_handles = tuple(
    Line2D([], [], marker=m, color='black', markerfacecolor='none',
           markeredgewidth=1.2, markersize=5, linestyle='none')
    for m in unique_markers
)

leg = ax1.legend(
    [exp_handles, sim_pts, fit_line],
    ['Experimental values', r'Geant4 simulated values',
     'Empirical fit'],
    handler_map={tuple: HandlerTuple(ndivide=None, pad=0.3)},
    handlelength=len(unique_markers) * 1.0,
    fontsize=11,
    framealpha=1,
    facecolor='white',
    edgecolor='black',
    fancybox=False,
    borderpad=0.8,
    labelspacing=0.7,
    handletextpad=0.8,
)
leg.get_frame().set_linewidth(0.8)
ax1.set_ylabel(r'$\epsilon_\mathrm{abs}$ (%)', fontsize=13)
ax1.set_ylim(0, .8)


# Bottom panel: residuals
ax2.axhspan(-2, 2, color='#F0F0F0', zorder=0)   # ±2σ
ax2.axhspan(-1, 1, color='#D9D9D9', zorder=0)   # ±1σ
ax2.axhline(0, color='black', linewidth=0.8)

for E, p, m in zip(energy_sorted, residuals, markers):
    ax2.plot(E, p, marker=m, color='black', markerfacecolor='none',
             markeredgewidth=1.2, markersize=9, linestyle='none')

ax2.set_xlabel(r'$E$ (keV)', fontsize=13)
ax2.set_ylabel(r'$(\epsilon_\mathrm{abs}^{\mathrm{exp}}-\epsilon_\mathrm{abs}^{\mathrm{MC}})/\delta\epsilon_\mathrm{abs}^\mathrm{exp}$',
               fontsize=13)

lim = max(3, np.ceil(np.abs(residuals).max()) + 0.5)
ax2.set_ylim(-lim, lim)
ax2.yaxis.set_major_locator(MaxNLocator(nbins=5, integer=True, symmetric=True))

ax2.grid(False, which='both')

ax2.set_xlim(0, 1400)
ax2.set_ylim(-3, 3)
fig.align_ylabels([ax1, ax2])
fig.savefig('abs_eff_plot.pdf', bbox_inches='tight', pad_inches=0.01)
plt.show()

for E, p, s in zip(energy_sorted, residuals, delta_abs_eff_sorted):
    print(f"{E:9.2f} keV   residual = {p:+6.2f} sigma   (sigma_exp = {s:.5f} %)")
```
